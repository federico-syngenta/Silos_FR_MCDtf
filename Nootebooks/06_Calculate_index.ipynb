```python
import os
import glob
import rasterio
import numpy as np

# --- CONFIGURACIÓN ---
# Directorio con tus imágenes originales de 4 bandas
INPUT_DIR = r"D:\Silos\Base de datos\procesado_sin_nubes\tiles"

# Directorio donde se guardarán los nuevos rasters del índice
OUTPUT_DIR = os.path.join(INPUT_DIR, "indices_INBNV")

# ¡IMPORTANTE! Asignación de bandas. Verifica que esto coincida con tus archivos.
# Para Sentinel-2, lo común es:
# BANDA_VERDE = 2
# BANDA_ROJA = 3
# BANDA_NIR = 4
# Si tus archivos tienen un orden distinto (ej. R,G,B,NIR -> 1,2,3,4), ajusta los números.
BANDA_VERDE = 2 
BANDA_ROJA = 1
BANDA_NIR = 4
# --- FIN DE LA CONFIGURACIÓN ---

# Crear el directorio de salida si no existe
os.makedirs(OUTPUT_DIR, exist_ok=True)

# Buscar todos los archivos .tif en el directorio de entrada
raster_files = glob.glob(os.path.join(INPUT_DIR, "*.tif"))

print(f"Se encontraron {len(raster_files)} imágenes para procesar.")

for file_path in raster_files:
    try:
        print(f"Procesando: {os.path.basename(file_path)}...")

        with rasterio.open(file_path) as src:
            # Leer las bandas necesarias como arrays de numpy
            # Convertimos a float32 para permitir cálculos decimales
            green = src.read(BANDA_VERDE).astype(np.float32)
            red = src.read(BANDA_ROJA).astype(np.float32)
            nir = src.read(BANDA_NIR).astype(np.float32)
            
            # Obtener los metadatos del archivo original para replicarlos
            meta = src.meta.copy()

        # Evitar divisiones por cero. numpy se encargará, pero suprimimos el warning.
        np.seterr(divide='ignore', invalid='ignore')

        # --- CÁLCULO DEL ÍNDICE INBNV ---
        # Numerador: (Verde + Rojo) - NIR
        # Denominador: (Verde + Rojo) + NIR
        numerator = (green + red) - nir
        denominator = (green + red) + nir
        
        inbnv = np.where(denominator == 0, 0, numerator / denominator) # Si el denominador es 0, el resultado es 0
        
        # Actualizar los metadatos para el nuevo archivo de salida
        meta.update({
            'dtype': 'float32', # El resultado es un número decimal
            'count': 1,         # El nuevo raster solo tiene una banda
            'nodata': -9999     # Un valor para representar "sin datos"
        })
        
        # Reemplazar posibles valores infinitos o NaN con el valor nodata
        inbnv[~np.isfinite(inbnv)] = meta['nodata']

        # Construir el nombre del archivo de salida
        base_name = os.path.basename(file_path)
        name_part, ext = os.path.splitext(base_name)
        output_filename = f"{name_part}_INBNV{ext}"
        output_path = os.path.join(OUTPUT_DIR, output_filename)

        # Guardar el nuevo raster
        with rasterio.open(output_path, 'w', **meta) as dst:
            dst.write(inbnv, 1)

    except Exception as e:
        print(f"ERROR al procesar {os.path.basename(file_path)}: {e}")

print("\n¡Proceso completado! 🚀")
print(f"Los índices se guardaron en: {OUTPUT_DIR}")
```

```python
import os
import glob
import rasterio
import numpy as np
from skimage.filters import threshold_otsu, threshold_triangle

# --- CONFIGURACIÓN ---
# Directorio donde están tus índices INBNV
INPUT_DIR = r"D:\Silos\Base de datos\procesado_sin_nubes\tiles\indices_INBNV"

# Método de umbralización: 'percentile' o 'triangle'
METHOD = 'percentile'  

# Solo si usas 'percentile', define el porcentaje
PERCENTILE = 99.8  # ejemplo: 99.8 = 0.2% de los píxeles más brillantes

# Directorio de salida para las máscaras
OUTPUT_DIR = os.path.join(INPUT_DIR, "masks")
os.makedirs(OUTPUT_DIR, exist_ok=True)
# --- FIN DE CONFIGURACIÓN ---

# Buscar todos los archivos INBNV
indice_files = glob.glob(os.path.join(INPUT_DIR, "*_INBNV.tif"))
print(f"Se encontraron {len(indice_files)} índices para procesar.")

for path_al_indice in indice_files:
    try:
        print(f"\nProcesando: {os.path.basename(path_al_indice)}")

        with rasterio.open(path_al_indice) as src:
            inbnv = src.read(1)
            nodata = src.nodata
            meta = src.meta.copy()

        # Filtrar solo los píxeles válidos
        valid_pixels = inbnv[inbnv != nodata]

        if valid_pixels.size == 0:
            print("⚠️ Imagen sin píxeles válidos, se salta.")
            continue

        # Calcular el umbral
        if METHOD == 'percentile':
            print(f"Usando Percentil {PERCENTILE}...")
            threshold = np.percentile(valid_pixels, PERCENTILE)
        elif METHOD == 'triangle':
            print("Usando método Triángulo...")
            threshold = threshold_triangle(valid_pixels)
        else:
            print("Método desconocido, usando Otsu como fallback...")
            threshold = threshold_otsu(valid_pixels)

        print(f"Umbral calculado: {threshold:.4f}")

        # Crear máscara binaria
        binary_mask = np.where(inbnv > threshold, 1, 0).astype(np.uint8)

        # Marcar nodata
        if nodata is not None:
            binary_mask[inbnv == nodata] = 255

        # Actualizar metadatos
        meta.update({'dtype': 'uint8', 'nodata': 255})

        # Guardar salida
        base_name = os.path.basename(path_al_indice).replace("_INBNV.tif", f"_{METHOD.upper()}_MASK.tif")
        output_mask_path = os.path.join(OUTPUT_DIR, base_name)

        with rasterio.open(output_mask_path, 'w', **meta) as dst:
            dst.write(binary_mask, 1)

        print(f"✅ Máscara guardada en: {output_mask_path}")

    except Exception as e:
        print(f"❌ ERROR con {os.path.basename(path_al_indice)}: {e}")

print("\n--- Proceso terminado --- 🚀")
print(f"Todas las máscaras están en: {OUTPUT_DIR}")
```

```python
import os
import glob
import numpy as np
import rasterio
from rasterio.features import shapes
from scipy.ndimage import label, binary_dilation
import geopandas as gpd
from shapely.geometry import shape

# --- CONFIGURACIÓN ---
MASK_DIR = r"D:\Silos\Base de datos\procesado_sin_nubes\tiles\indices_INBNV\masks"
OUTPUT_DIR = os.path.join(MASK_DIR, "polygons")
os.makedirs(OUTPUT_DIR, exist_ok=True)

# Parámetros
MIN_PIXELS = 3   # eliminar objetos con <= 3 píxeles
BUFFER_PIXELS = 1  # sumar 1 píxel de buffer
# --- FIN DE CONFIGURACIÓN ---

mask_files = glob.glob(os.path.join(MASK_DIR, "*_MASK.tif"))
print(f"Se encontraron {len(mask_files)} máscaras para procesar.")

for mask_path in mask_files:
    try:
        print(f"\nProcesando: {os.path.basename(mask_path)}")

        with rasterio.open(mask_path) as src:
            mask = src.read(1)
            transform = src.transform
            crs = src.crs

        # Solo los píxeles válidos (valor 1)
        binary = (mask == 1).astype(np.uint8)

        # --- 1. Eliminar objetos pequeños ---
        labeled, num_features = label(binary)
        cleaned = np.zeros_like(binary)

        for i in range(1, num_features + 1):
            region = (labeled == i)
            if region.sum() > MIN_PIXELS:
                cleaned[region] = 1

        # --- 2. Buffer de 1 píxel ---
        buffered = binary_dilation(cleaned, iterations=BUFFER_PIXELS).astype(np.uint8)

        # --- 3. Raster a polígonos ---
        polygons = []
        for geom, val in shapes(buffered, mask=buffered.astype(bool), transform=transform):
            if val == 1:  # solo áreas válidas
                polygons.append(shape(geom))

        if not polygons:
            print("⚠️ No quedaron polígonos válidos.")
            continue

        # --- 4. Guardar a Shapefile ---
        gdf = gpd.GeoDataFrame(geometry=polygons, crs=crs)
        output_file = os.path.join(
            OUTPUT_DIR,
            os.path.basename(mask_path).replace("_MASK.tif", "_POLYGONS.shp")
        )
        gdf.to_file(output_file)

        print(f"✅ Polígonos guardados en: {output_file}")

    except Exception as e:
        print(f"❌ ERROR con {os.path.basename(mask_path)}: {e}")

print("\n--- Proceso terminado --- 🚀")
print(f"Todos los polígonos están en: {OUTPUT_DIR}")
```
